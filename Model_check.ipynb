```python
import numpy as np
import pandas as pd
import pickle
```

```python
vec = pickle.load(open('vectorizer.pkl','rb'))
model = pickle.load(open('model.pkl','rb'))

df = pd.read_csv("spam.csv",encoding="latin1")
df.head()
```

```python
df = df.rename(columns={'v1':"label",'v2':'Message'})
df = df[["label","Message"]]
df.head()
```

```python
X_test = vec.transform(df['Message'])
Y_test = df['label']
```

```python
Y_pred = model.predict(X_test)
```

```python

Y_test_numeric = Y_test.map({'ham': 0, 'spam': 1})
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,precision_score

print(f"Accuracy: {accuracy_score(Y_test_numeric, Y_pred)}")
print("Classification Report: ")
print(classification_report(Y_test_numeric,Y_pred))
print()
print("Confusion Matrix: ")
print(confusion_matrix(Y_test_numeric, Y_pred))
print(f"precision score: {precision_score(Y_test_numeric, Y_pred)}")
```
